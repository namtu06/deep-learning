## Quick Overview: Gradient Descent

This section will review how forward and backward passes work and how parameters are updated during training.

### 1. Forward pass

The forward pass involves computing the output of each layer from its inputs by traversing through each block of the neural network from the first to the last layer.

For each fully connected (linear) layer, the output is calculated using its input vector, weights, and biases as follows:

$$
z^{(l)} = W^{(l)}a^{(l-1)} + b^{(l)}. 
$$

Here, $W^{(l)}$ and $b^{(l)}$ are the weights and the biases of the $l^{th}$ layer, respectively, while $a^{(l-1)}$ is the output from the previous layer.  For the first layer, $a^{(0)}$ represents the input to the network, commonly denoted as $x$. The output of the linear layer, $z^{(l)}$, is often referred to as the pre-activation. Next, we  apply a nonlinearity to compute the post-activation output:

$$
a^{(l)} = f(z^{(l)}),
$$
where $f$ represents the activation function (such as tanh, ReLU, etc.).

Once the output layer produces a prediction $y$, the loss function is computed. In this case, we are using the mean squared error (MSE) loss:

$$
L = \frac{1}{N} \sum_{n=1}^N (y_n - t_n)^2,
$$

where $y_n$ is the predicted value for data point $n$, $t_n$ is its target value, and $N$ is the number of data points.


### 2. Backward pass

The backward pass involves computing the gradients of the loss with respect to the model parameters (weights and biases) of each layer by traversing from the last layer to the first layer. This process uses the chain rule to propagate the error gradients backward through the network and is also known as **backpropagation**. During the backward pass, we compute the gradients for both the pre-activation and the post-activation outputs.


### 3. Update parameters

Once the gradients are calculated, the model parameters are updated using the gradient descent algorithm. The parameters are adjusted in the direction that reduces the loss function. This update step is performed using the following rule:

$$
\theta^{(\tau)} = \theta^{(\tau-1)} - \eta \cdot \nabla_{\theta}L^{(\tau-1)},
$$


where $\theta$ represents the parameters (weights and biases), $\eta$ is the learning rate, $\tau$ labels the iteration step, and $\nabla_{\theta}L$ is the gradient of the loss function with respect to the parameters.

In practice, training a neural network involves multiple iterations over the dataset. Each pass over the whole dataset is called an epoch. During each epoch, the dataset is typically divided into smaller subsets called batches. The model parameters are updated after each batch. This approach, known as mini-batch gradient descent, helps in efficient computation and faster convergence. However, for this assignment, we will perform the computation for only one iteration (one forward pass, one backward pass, and parameter update).

### How to Derive the Gradients: Element-wise First, Then Matrix Form

It is tempting to write the chain rule directly with vectors and matrices, e.g. $\frac{\partial L}{\partial W_2} = \frac{\partial L}{\partial y}\frac{\partial y}{\partial W_2}$. However, $\frac{\partial y}{\partial W_2}$ is the derivative of a vector with respect to a matrix, which is a three-dimensional array. It is not obvious in which order the factors should be multiplied or where the transposes go.

A more reliable way is to work with **individual elements**. Every element is a scalar, so you only need the ordinary chain rule from scalar calculus. Once you have the formula for one element, you rewrite it as a matrix operation. The matrix form is what you will implement in NumPy in Part 2.

**Notation.** We use the following indices:
- $k = 1, 2, 3$ for the inputs $x_k$,
- $j = 1, 2$ for the hidden neurons ($z_j$, $a_j$),
- $i = 1$ for the output neuron ($y_i$, $t_i$). There is only one output here, but keeping the index $i$ makes the formulas valid for any number of outputs.

$(W_1)_{jk}$ is the element in row $j$ and column $k$ of $W_1$, and $(b_1)_j$ is the $j$-th element of $b_1$. For any vector or matrix $P$, $\frac{\partial L}{\partial P}$ denotes the array **with the same shape as $P$** whose elements are $\frac{\partial L}{\partial P_{jk}}$. For example, $\frac{\partial L}{\partial W_1}$ is a $2 \times 3$ matrix, just like $W_1$. This is also the shape the answer cells expect.

**The forward pass, one element at a time:**

$$
\begin{aligned}
z_j &= \sum_{k=1}^{3} (W_1)_{jk}\, x_k + (b_1)_j, \\
a_j &= \tanh(z_j), \\
y_i &= \sum_{j=1}^{2} (W_2)_{ij}\, a_j + (b_2)_i, \\
L &= (y_1 - t_1)^2.
\end{aligned}
$$

Figure 2 draws the same network neuron by neuron. Each arrow is one term of the equations above. For example, the arrow $x_k \to z_j$ carries the weight $(W_1)_{jk}$: the row index $j$ is the neuron the arrow goes into, and the column index $k$ is the input it comes from. The highlighted path shows that the weight $(W_1)_{12}$ affects the loss only through $z_1 \to a_1 \to y_1 \to L$.

**Procedure for each gradient:**

1. **Find the paths.** Use the element-wise equations above, or Figure 2, to see where the variable is used. If it is used in only one quantity, the chain rule is a single product. For example, $(W_2)_{ij}$ is used only in $y_i$, so $\frac{\partial L}{\partial (W_2)_{ij}} = \frac{\partial L}{\partial y_i}\frac{\partial y_i}{\partial (W_2)_{ij}}$. If it is used in several quantities, sum over all of them: if a variable $u$ is used in $v_1, \dots, v_n$, then $\frac{\partial L}{\partial u} = \sum_{m=1}^{n} \frac{\partial L}{\partial v_m}\frac{\partial v_m}{\partial u}$.

2. **Compute the local derivatives** from the element-wise equations. When you differentiate a sum such as $\sum_{j} (W_2)_{ij}\, a_j$ with respect to one of its elements, only the term that contains that element remains. All other terms are constants, and their derivative is zero.

3. **Convert the result to matrix form** by looking at the indices:

| Element-wise result | Matrix form | Shapes |
|---|---|---|
| $C_{jk} = u_j\, v_k$ | $C = u\, v^T$ (outer product) | $(m \times 1)(1 \times n) = m \times n$ |
| $c_j = \sum_i M_{ij}\, u_i$ (sum over the **row** index of $M$) | $c = M^T u$ | $(n \times m)(m \times 1) = n \times 1$ |
| $c_i = \sum_j M_{ij}\, u_j$ (sum over the **column** index of $M$) | $c = M u$ | $(m \times n)(n \times 1) = m \times 1$ |
| $c_j = u_j\, v_j$ | $c = u \odot v$ (element-wise product, `*` in NumPy) | same shape as $u$ and $v$ |

4. **Check the shape.** $\frac{\partial L}{\partial P}$ must have the same shape as $P$. If the shapes do not match, a transpose is usually missing. Go back to the element-wise formula to see where it belongs.

Step 3.2 below shows this procedure for $\frac{\partial L}{\partial W_2}$. Use the same procedure for the other gradients.

> **Note on notation:** As in the lecture notes *Gradient calculation in multilayer perceptrons*, $z$ is the pre-activation and $a$ is the activation. The lecture notes use different symbols for the rest: $\hat{y}$ for the prediction, $y$ for the target and $c$ for the loss, where this notebook writes $y$, $t$ and $L$. The weight indices are also in a different order. The lecture notes write $w_{ij}$ for the weight from input $i$ to neuron $j$, in the order the signal flows. A weight matrix therefore has one row per neuron and one column per input; for example, $W_1$ is $2 \times 3$. Part 2 and PyTorch store weights in the same way.

```python
# Answer format checker. DO NOT EDIT OR DELETE THIS CELL.
# Run this cell once before you start, and again after every kernel restart.
# It defines check_answer_format() and check_all_answer_formats(), which check the FORMAT
# of your answers (brackets, number of values, numbers written correctly).
# They do not check whether your values are correct, and they give no points themselves.
import numbers
import textwrap
import numpy as np

# The form of each answer, as in its template: () = a single number,
# (n,) = a list of n numbers, (rows, columns) = a list of lists.
ANSWER_SHAPES = {
    'z': (2,), 'a': (2,), 'y': (1,), 'loss': (),
    'dL_dy': (1,), 'dL_dW2': (1, 2), 'dL_db2': (1,), 'dL_dW1': (2, 3),
    'W2_updated': (1, 2), 'b2_updated': (1,), 'W1_updated': (2, 3), 'b1_updated': (2,),
}

COMMA_HINT = ('Did you write a comma as the decimal point? Python reads 0,45 as two numbers, 0 and 45. '
              'Write the decimal point as a dot: 0.45.')


class AnswerFormatError(AssertionError):
    """An answer does not have the format of its template."""

    def _render_traceback_(self):
        # Jupyter shows only this message, without the code of the checker.
        return ['\x1b[0;31mAnswerFormatError\x1b[0m: ' + str(self)]


def _template(shape):
    """The template of an answer, e.g. (1, 2) -> '[[None, None]]'.

    The spaces are non-breaking, so that a long message is never wrapped inside the template.
    _bullets() turns them back into normal spaces.
    """
    if len(shape) == 0:
        return 'None'
    row = '[' + ',\xa0'.join(['None'] * shape[-1]) + ']'
    if len(shape) == 1:
        return row
    return '[' + ',\xa0'.join([row] * shape[0]) + ']'


def _count(n, word):
    return f'{n} {word}' if n == 1 else f'{n} {word}s'


def _join(items):
    """['a', 'b', 'c'] -> 'a, b and c'."""
    return items[0] if len(items) == 1 else ', '.join(items[:-1]) + ' and ' + items[-1]


def _is_list(value):
    return isinstance(value, (list, tuple, np.ndarray))


def _not_filled(name, template):
    return (f'{name} is not filled in: it is still {template}, as in its template. '
            f'Replace every None with a number and run the cell. If you have done that, look for '
            f'an error message under your answer cell: the cell stopped before it stored your answer.')


def _not_defined(name):
    return (f'{name} is not defined, so the cell with your answer has not run without errors. Run it, and '
            f'if it shows an error, see the table of error messages in "Check the Format of Your Answers".')


def _shape_problems(name, value, shape, template):
    """Problems with the brackets or the number of values. Empty if the shape is right."""
    if shape == ():
        if not _is_list(value):
            return []
        problem = (f'{name} must be a single number without brackets, e.g. {name} = 0.123, '
                   f'but it is a {type(value).__name__} of {_count(len(value), "value")}.')
        if isinstance(value, tuple):
            problem += ' ' + COMMA_HINT
        return [problem]

    if not _is_list(value):
        kind = 'a single number' if isinstance(value, numbers.Number) else f'a {type(value).__name__}'
        return [f'{name} must be a list like its template {template}, but it is {kind}. '
                f'Keep the brackets of the template.']

    if len(shape) == 1:
        n = shape[0]
        problems = []
        if any(_is_list(v) for v in value):
            problems.append(f'{name} must be a flat list of numbers like its template {template}, '
                            f'without inner brackets.')
        if len(value) != n:
            problem = (f'{name} must contain {_count(n, "number")} like its template {template}, '
                       f'but it contains {len(value)}.')
            if len(value) > n:
                problem += ' ' + COMMA_HINT
            problems.append(problem)
        return problems

    n_rows, n_cols = shape
    if not any(_is_list(row) for row in value):
        problem = (f'{name} must be a list of lists (note the double brackets) like its template '
                   f'{template}, but it is a flat list.')
        if n_rows == 1 and len(value) == n_cols:
            problem += f' Add the outer brackets: {name} = [[...]].'
        return [problem]
    if not all(_is_list(row) for row in value):
        return [f'Every row of {name} must be a list like in its template {template}.']
    lengths = [len(row) for row in value]
    if len(value) == n_rows and all(length == n_cols for length in lengths):
        return []
    if len(value) == n_cols and all(length == n_rows for length in lengths):
        problem = (f'{name} has {_count(n_cols, "row")} of {_count(n_rows, "number")}, but it must have '
                   f'{_count(n_rows, "row")} of {_count(n_cols, "number")} like its template {template}. '
                   f'Did you transpose it?')
        if name.startswith('dL_'):
            problem += ' A gradient has the same shape as its parameter.'
        return [problem]
    problems = []
    if len(value) != n_rows:
        problems.append(f'{name} must have {_count(n_rows, "row")} like its template {template}, '
                        f'but it has {len(value)}.')
    for i, length in enumerate(lengths):
        if length != n_cols:
            problems.append(f'Row {name}[{i}] must contain {_count(n_cols, "number")}, '
                            f'but it contains {length}.')
    if any(length > n_cols for length in lengths):
        problems.append(COMMA_HINT)
    return problems


def _format_problems(name, value, shape):
    """Everything that is wrong with the format of an answer. Empty if the format is right."""
    template = _template(shape)
    if isinstance(value, np.ndarray) and value.ndim == 0:
        value = value.item()
    if value is None:
        return [_not_filled(name, template)]
    problems = _shape_problems(name, value, shape, template)
    if problems:
        return problems

    # The brackets are right; now check each value.
    if shape == ():
        values = [(name, value)]
    elif len(shape) == 1:
        values = [(f'{name}[{i}]', v) for i, v in enumerate(value)]
    else:
        values = [(f'{name}[{i}][{j}]', v) for i, row in enumerate(value) for j, v in enumerate(row)]
    nones = [pos for pos, v in values if v is None]
    if len(nones) == len(values):
        return [_not_filled(name, template)]
    if nones:
        problems.append(f'{_join(nones)} {"is" if len(nones) == 1 else "are"} still None. '
                        f'Replace every None with a number.')
    whole = []
    for pos, v in values:
        if v is None:
            continue
        if isinstance(v, (bool, np.bool_)):
            problems.append(f'{pos} is {v}, not a number.')
        elif isinstance(v, str):
            problems.append(f'{pos} is the text {v!r}, not a number. Write the number without '
                            f'quotes, with a dot as the decimal point, e.g. 0.45.')
        elif not isinstance(v, numbers.Real):
            problems.append(f'{pos} is not a number: Python reads it as {v!r} ({type(v).__name__}).')
        elif not np.isfinite(v):
            problems.append(f'{pos} is {v}, not a finite number.')
        elif isinstance(v, numbers.Integral) and v != 0:
            whole.append(f'{pos}\xa0=\xa0{v}')
    if whole:
        problems.append(
            f'{_join(whole)} {"is a whole number" if len(whole) == 1 else "are whole numbers"}, '
            f'but no answer in this exercise is a whole number. Did you write a comma as the decimal point? '
            f'Python reads 0,45 as two numbers, 0 and 45. '
            f'Write the decimal point as a dot: 0.45. (If you really mean a whole number, write it as 1.0.)')
    return problems


def _bullets(problems, indent):
    # Wrap the lines, then turn non-breaking spaces back into normal ones, so that a
    # template copied from a message is valid Python.
    text = '\n'.join(textwrap.fill(p, width=100, initial_indent=indent + '- ',
                                   subsequent_indent=indent + '  ') for p in problems)
    return text.replace('\xa0', ' ')


def check_answer_format(name):
    """Check the format of the answer with this name. Raise AnswerFormatError if it is wrong."""
    if name not in globals():
        raise AnswerFormatError(textwrap.fill(_not_defined(name), width=100))
    value = globals()[name]
    problems = _format_problems(name, value, ANSWER_SHAPES[name])
    if problems:
        raise AnswerFormatError(
            f'{name} does not have the right format.\n'
            f'Python reads your answer as: {name} = {value!r}\n'
            f'{_bullets(problems, "")}\n'
            f'Fix your answer, run its cell, and then run this check again.')
    print(f'Format OK: {name} = {value!r}')
    print('This checks only the format, not whether the values are correct.')


def check_all_answer_formats():
    """Check the format of all answers and list every problem."""
    answers = globals()
    n_wrong = 0
    for name, shape in ANSWER_SHAPES.items():
        if name not in answers:
            value_text = '(not defined)'
            problems = [_not_defined(name)]
        else:
            value_text = repr(answers[name])
            problems = _format_problems(name, answers[name], shape)
        print(f'{name:<11} {"PROBLEM" if problems else "OK":<8} {value_text}')
        if problems:
            n_wrong += 1
            print(_bullets(problems, ' ' * 12))
    if n_wrong:
        raise AnswerFormatError(
            f'{n_wrong} of {len(ANSWER_SHAPES)} answers have a format problem (see the list above). '
            f'Fix them, run their cells, and then run this check again.')
    print(f'\nAll {len(ANSWER_SHAPES)} answers have the right format. '
          f'This checks only the format, not whether the values are correct.')
```

##### 2. Gradient of the Loss w.r.t the Output Layer Weights:

This step is a worked example of the element-wise procedure.

**Element-wise.** The weight $(W_2)_{ij}$ appears only in the output $y_i = \sum_{j'} (W_2)_{ij'}\, a_{j'} + (b_2)_i$. There is a single path from $(W_2)_{ij}$ to $L$:

$$
\frac{\partial L}{\partial (W_2)_{ij}} = \frac{\partial L}{\partial y_i}\,\frac{\partial y_i}{\partial (W_2)_{ij}}.
$$

In the sum over $j'$, only the term $j' = j$ contains $(W_2)_{ij}$, so $\frac{\partial y_i}{\partial (W_2)_{ij}} = a_j$ and

$$
\frac{\partial L}{\partial (W_2)_{ij}} = \frac{\partial L}{\partial y_i}\, a_j.
$$

**Matrix form.** Element $(i, j)$ is the $i$-th element of $\frac{\partial L}{\partial y}$ times the $j$-th element of $a$. This is the outer-product pattern $C_{ij} = u_i\, v_j \Rightarrow C = u\, v^T$:

$$
\frac{\partial L}{\partial W_2} = \frac{\partial L}{\partial y}\, a^T = \begin{bmatrix} \frac{\partial L}{\partial y_1} a_1 & \frac{\partial L}{\partial y_1} a_2 \end{bmatrix}.
$$

**Shape check:** $(1 \times 1)(1 \times 2) = 1 \times 2$, the same as $W_2$.
